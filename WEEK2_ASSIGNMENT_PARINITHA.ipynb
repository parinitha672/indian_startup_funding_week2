```python
#CLEANING CODE
import pandas as pd
import numpy as np
# Load dataset
df = pd.read_csv("indian_startups_fundingDADS.csv")
# Clean column names
df.columns = (
    df.columns.str.strip()
              .str.lower()
              .str.replace(" ", "_")
)
# Rename columns
df.columns = [
    "startup_name",
    "city",
    "industry",
    "funding_stage",
    "amount",
    "funding_date",
    "investors",
    "founded_year"
]
# Clean startup name and city
df["startup_name"] = df["startup_name"].str.strip().str.title()
df["city"] = df["city"].str.strip().str.title()
# Standardize city names
city_map = {
    "Bangalore": "Bengaluru",
    "Blr": "Bengaluru",
    "Bombay": "Mumbai",
    "Gurgaon": "Gurugram",
    "Delhi": "New Delhi",
    "Calcutta": "Kolkata",
    "Madras": "Chennai"
}
df["city"] = df["city"].replace(city_map)
# Standardize industry
industry_map = {
    "Fin Tech": "Fintech",
    "Fintech ": "Fintech",
    "Ed Tech": "Edtech",
    "Health Tech": "HealthTech",
    "E-Commerce": "Ecommerce",
    "Ai": "AI"
}
df["industry"] = df["industry"].replace(industry_map)
# Standardize funding stage
stage_map = {
    "Seed Round": "Seed",
    "Pre Seed": "Pre-Seed",
    "Series A Round": "Series A",
    "Series B Round": "Series B",
    "Series C Round": "Series C"
}
df["funding_stage"] = df["funding_stage"].replace(stage_map)
# Clean amount column
df["amount"] = (
    df["amount"]
      .astype(str)
      .str.replace("₹","",regex=False)
      .str.replace("INR","",regex=False)
      .str.replace("Cr","",regex=False)
      .str.replace("Crore","",regex=False)
      .str.replace(",","",regex=False)
      .str.strip()
      .replace("Undisclosed", np.nan)
)
df["amount"] = pd.to_numeric(df["amount"], errors="coerce")
# Convert funding date
df["funding_date"] = pd.to_datetime(
    df["funding_date"],
    errors="coerce",
    dayfirst=True
)
# Remove duplicate rows
df = df.drop_duplicates()
```

Q34. For each city, compute three metrics in a single groupby using .agg() with a dictionary:
(a) Number of startups
(b) Total funding raised (sum of amount)
(c) Average funding per startup
Print the top 10 cities by number of startups.

```python
city_summary = (
    df.groupby("city")
      .agg({
          "startup_name": "count",
          "amount": ["sum", "mean"]
      })
)
city_summary.columns = [
    "startup_count",
    "total_funding",
    "average_funding"
]
city_summary = city_summary.sort_values(
    by="startup_count",
    ascending=False
)
print(city_summary.head(10))
```

Q35. Interview-style. Which industry has the highest average funding, but only considering industries
with at least 15 startups? (Sparse industries bias the mean.) Print industry name, count, and average
funding.

```python
industry_stats = (
    df.groupby("industry")
      .agg(
          count=("startup_name", "count"),
          average_funding=("amount", "mean")
      )
)
industry_stats = industry_stats[
    industry_stats["count"] >= 15
]
industry_stats = industry_stats.sort_values(
    by="average_funding",
    ascending=False
)
print(industry_stats.head(1))
```

Q39. Investor hub analysis. A foreign investor wants to know which Indian city has the highest TOTAL
funding raised, considering only cities with at least 20 startups (so the number is meaningful). Print city
name, count, and total funding.

```python
city_data = (
    df.groupby("city")
      .agg(
          startup_count=("startup_name", "count"),
          total_funding=("amount", "sum")
      )
)
city_data = city_data[
    city_data["startup_count"] >= 20
]
city_data = city_data.sort_values(
    by="total_funding",
    ascending=False
)
print(city_data.head(1))
```

Q40. Market saturation. For each combination of (city, industry), count the number of startups. Print the
top 10 most-saturated combinations. This tells a new founder where competition is most intense – and
by elimination, where the gaps might be.

```python
market = (
    df.groupby(["city", "industry"])
      .size()
      .reset_index(name="startup_count")
)
market = market.sort_values(
    by="startup_count",
    ascending=False
)
print(market.head(10))
```
